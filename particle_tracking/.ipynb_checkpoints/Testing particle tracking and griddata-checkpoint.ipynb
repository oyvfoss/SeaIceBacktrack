```python
lat_ctr, lon_ctr = 86, 30

cart_proj = ccrs.Stereographic(
    central_latitude=lat_ctr,
    central_longitude=lon_ctr,
    scale_factor=None, globe=None)

fig = plt.figure(figsize = (8, 8))
ax = plt.axes(projection=cart_proj)

ax.set_facecolor('skyblue'  )
# Coastline
ax.coastlines('50m', color = 'w', lw = 0.5 )

# Land
land_50m = cfeature.NaturalEarthFeature(
    'physical', 'land', '50m',
    edgecolor='k', linewidth = 1,
    facecolor='lightgray')

ax.add_feature(land_50m)

# Set boundaries
ax.set_extent([-20, 191, 66, 89.5], crs=ccrs.PlateCarree())


ax.plot(P.lon, P.lat, 'k',  transform=ccrs.PlateCarree(), lw = 0.5)
ax.plot(P.lon0, P.lat0, '*r', transform=ccrs.PlateCarree())
ax.plot(P.lon_origin, P.lat_origin, '*', color = (1,1,0), transform=ccrs.PlateCarree())
#ax.plot(P2.lon, P2.lat, 'y', transform=ccrs.PlateCarree())
#ax.plot(P2.lon0, P2.lat0, '*r', transform=ccrs.PlateCarree())
#ax.plot(P2.lon0, P2.lat0, '*y', transform=ccrs.PlateCarree())
```
