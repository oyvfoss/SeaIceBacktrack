## Define particle tracking class
Defining a class *IceParticle**

```python
import numpy as np
import glob
import pyproj
import xarray as xr
from scipy.interpolate import griddata
from scipy.interpolate import RegularGridInterpolator
from matplotlib.dates import date2num, num2date
import datetime
import pandas as pd
from scipy.spatial import cKDTree
```

## Utility functions 

```python
def closest(x, x0):
    '''
    Takes a vector x and returns the index for which x is closest to a
    value x0
    '''
    dx = abs(x-x0)
    ind = np.nonzero(dx==np.ma.min(dx))[0]
    return ind[0]
```

```python
class IceParticleInputs:
    """Shared grid, KD-tree, masks, velocity arrays."""
    def __init__(self, ds,):
        self.ds = ds
        self.u_all = ds.u
        self.v_all = ds.v

        self.land_mask_PP = np.isnan(ds.u.mean(dim='time').values)
        self.xg, self.yg = ds.x.values, ds.y.values
        self.X, self.Y = np.meshgrid(ds.x.data, ds.y.data)
        ocean_mask = ~self.land_mask_PP
        self.ny, self.nx = self.X.shape
        self.X_flat = self.X.ravel()
        self.Y_flat = self.Y.ravel()
        self.ocean_flat_idx = np.flatnonzero(ocean_mask)

        i_coords, j_coords = np.meshgrid(np.arange(self.ny), np.arange(self.nx), indexing='ij')
        self.ocean_ij = np.column_stack([
            i_coords.ravel()[self.ocean_flat_idx],
            j_coords.ravel()[self.ocean_flat_idx]
        ])
        self.ocean_tree = cKDTree(np.column_stack([self.X_flat[self.ocean_flat_idx],
                                                   self.Y_flat[self.ocean_flat_idx]]))
```

```python
class IceParticle:

    # Define some class variables (projection)
    ease2_proj = pyproj.CRS("+proj=laea +lat_0=90 +lon_0=0 +x_0=0 +y_0=0 " \
                  + "+a=6371228 +b=6371228 +units=m +no_defs") # NSIDC PP
    wgs84_proj = pyproj.CRS("EPSG:4326") # Lat-lon/WGS84 datum
    lonlat_to_ease2 = pyproj.Transformer.from_crs(
        wgs84_proj, ease2_proj, always_xy=True
    )
    
    ease2_to_lonlat = pyproj.Transformer.from_crs(
        ease2_proj, wgs84_proj, always_xy=True
    )

    
    def __init__(self, input_dict):
        '''
        Set up the IceParticle object.
    

        Inputs:
        -------
        
        *input_dict* should be on the form:
        
        {'ds': [xarray.Dataset containing daily NSIDC PP data], 
         'lat0', 'lon0': Location from where to run back-trajectory 
         'time0': Time to initialize back-trajectory (datetime object)
         'MASK': Xarray object with land mask or "None" (see 
                 "Load and prepare data.ipynb"and note below) 
         'interp_method': Interpolation method for *griddata" (e.g. 
                          "bilinear")
         'end_criteria': A string defining which of the end criteria below 
                         to use (for deciding when to stop  integration). 
                         E.g., "ABC" means using all the criteria, "AC"
                         just criteria A and C, etc.
         
                         Default (if not specified): *ABC"
         
            A. The particle is in open water (nearest non-landmasked 
               point has u = v = NaN (<15%SIC) 
            B. The particle has not moved (|u|<100 m /day for 3 
               consecutive days).
            C. The particle has drifted into land (defined by 
               MASK.is_land)  }
                          
        ---------------------------------------------------------------

        *MASK* should be an xarray Dataset with the following variables:
        
        - "x", "y" - Grid
        - "is_land" - Boolean (land=1, not land = 0)
        - "proj" - pyproj.CRS defining the grid projection of x, y

        '''


        self.inputs = inputs
        self.ds = inputs.ds
        #self.ds = self.ds.assign_coords(time=pd.to_datetime(self.ds.time.values))
        self.ds = ds.assign_coords(time=[pd.Timestamp(t).to_pydatetime() for t in self.ds.time.values])

        self.u_all = inputs.u_all
        self.v_all = inputs.v_all
        self.ocean_tree = inputs.ocean_tree
        self.ocean_ij = inputs.ocean_ij
        self.X = inputs.X
        self.Y = inputs.Y
        self.X_flat = inputs.X_flat
        self.Y_flat = inputs.Y_flat
        #self.X_ocean_flat = field.X_ocean_flat
        #self.Y_ocean_flat = field.Y_ocean_flat
        self.land_mask_PP = inputs.land_mask_PP
        self.ocean_flat_idx = inputs.ocean_flat_idx
        
        #print('Setting up object ..\r')
        
        # Set input_dict contents as object attributes
        for key in [ 'MASK', 'interp_method']:
            setattr(self, key, input_dict[key])
        
        self.lat0, self.lon0 = input_dict['lat0'], input_dict['lon0']
        self.x0, self.y0 = IceParticle.lonlat_to_ease2.transform(
            self.lon0, self.lat0)
        
        # Get the product land mask (stuff that is always nan)
      #  self.land_mask_PP = np.isnan(self.ds.u.mean(dim='time').values)

        # Experimental
       # self.u_all = ds.u.values 
       # self.v_all = ds.v.values 
        
        # 2D grid
        #self.X, self.Y = np.meshgrid(self.ds.x.data, self.ds.y.data)
        
        # Ocean mask (True for ocean points)
        ocean_mask = ~self.land_mask_PP

        # Grid shape
        self.ny, self.nx = self.X.shape
                
        # Find start index and start time
        self.tind0 = closest(date2num(input_dict['time0']), 
                             date2num(self.ds.time.data))
        self.time0 = self.ds.time.data[self.tind0]

        # Set up variables that we fill fill as we propagate 
        # the particle backward 
        self.time_num = [date2num(self.time0)]
        self.x = [self.x0]
        self.y = [self.y0]

                        
        # Drift vector components in the x (u) and y (v) grid directions 
        self.is_ice_initial, u0, v0 = self.get_uv(self.tind0, self.x0, self.y0) 
        self.u = [u0]
        self.v = [v0]

        
        # End criteria
        valid_end_crits = ['A', 'B', 'C', 'AB', 'AC', 'BC', 'ABC']
        if 'end_criteria' in input_dict.keys() == False:
            self.end_criteria = 'ABC'
        else:
            endcrit_in = input_dict['end_criteria']
            if endcrit_in == None:
                self.end_criteria = 'ABC'
            else:
                endcrit_sort = ''.join(sorted(endcrit_in)).upper()
                if endcrit_sort not in valid_end_crits:
                    raise Exception('Invalid end_criteria string: %s.'%endcrit_in,
                    '\nValid options are %s and None.'%valid_end_crits )
                self.end_criteria = endcrit_sort
        
        # Coordinate transformer functions
        self.ease2_to_landmask = pyproj.Transformer.from_crs(
            IceParticle.ease2_proj, self.MASK['proj'].item())

       # print(' ..done.\r')


    def get_uv(self, tind, x, y):
        """
        Returns: is_ice (bool), u (float), v (float)
        """

        
        # Query KD-tree for nearest 4 ocean points
        dists, idxs = self.ocean_tree.query([x, y], k=4)
    
        # Map to full-grid i,j indices
        ij_neighbors = self.ocean_ij[idxs]  # shape (4,2)
        i_neighbors, j_neighbors = ij_neighbors[:,0], ij_neighbors[:,1]
           
       # u_vals = np.array([self.u_all[tind, i, j] for i, j in zip(i_neighbors, j_neighbors)])
       # v_vals = np.array([self.v_all[tind, i, j] for i, j in zip(i_neighbors, j_neighbors)])

        u_vals = np.array([ds.u.isel(time=tind, y=i, x=j).compute().item() for i,j in zip(i_neighbors, j_neighbors)])
        v_vals = np.array([ds.v.isel(time=tind, y=i, x=j).compute().item() for i,j in zip(i_neighbors, j_neighbors)])

        
        # Check which neighbors are ice (not land & not NaN)
        ice_mask = ~np.isnan(u_vals)
        
        if np.sum(ice_mask) == 0:
            return False, np.nan, np.nan  # out of ice
        
        # Keep only ice neighbors
        u_vals = u_vals[ice_mask]
        v_vals = v_vals[ice_mask]
        i_neighbors = i_neighbors[ice_mask]
        j_neighbors = j_neighbors[ice_mask]
        
        # For coordinates, just grab the X/Y for those neighbors (flattened)
        coords = np.column_stack([self.X[i_neighbors, j_neighbors],
                                  self.Y[i_neighbors, j_neighbors]])

        # Distances
        dists_ice = np.linalg.norm(coords - np.array([x, y]), axis=1)
    
        # Weighted average or direct use if exactly on a grid point
        if np.any(dists_ice == 0):
            u = u_vals[dists_ice == 0][0]
            v = v_vals[dists_ice == 0][0]
        else:
            weights = 1 / dists_ice
            u = np.sum(weights * u_vals) / np.sum(weights)
            v = np.sum(weights * v_vals) / np.sum(weights)
    
        # Check ice presence at nearest ocean point
        idx_nearest = idxs[0]  # nearest neighbor from KD-tree
        i_near, j_near = self.ocean_ij[idx_nearest]
       # is_ice = not np.isnan(self.ds.u.isel(time=tind, y=i_near, x=j_near).values)
        is_ice = not np.isnan(self.u_all[tind, i_near, j_near])

        return is_ice, float(u), float(v)

    

    def is_on_land(self, x_ease, y_ease):
        '''
        Check if the lon, lat is within the land mask defined in MASK
        '''
        x_ib, y_ib = self.ease2_to_landmask.transform(x_ease, y_ease)

        is_land = np.round(self.MASK.is_land.interp(
            x=x_ib, y=y_ib, method = 'nearest').values)>0

        return is_land
    
                          

    def backtrack(self):
        
        self.end_reason = 'Error or user interrupt' 
        
        tind_ = self.tind0
        nn = 0

        out_of_ice = False
        
        if not self.is_ice_initial:
            out_of_ice = True
            self.end_reason = 'Started in open water.' 
            print('STARTED IN OPEN WATER!')

        
        # Loop backward in time until:
        # - A. The particle is in open water (nearest non-landmasked 
        #      point has u = v = NaN (<15%SIC) 
        # - B. The particle has not moved (|u|<100 m /day for 3 consecutive days).
        # - C. The particle has drifted into land (defined by MASK.is_land)      
                          
        while out_of_ice == False:
                          
            nn +=1
                          
            #### Propagating backwards #################
            # Step back the time index by 1              
            tind_ = tind_-1

            # Stop if we've reached beyond the start of the dataset
            if tind_ < 0:
                out_of_ice = True
                self.end_reason = "Reached beginning of dataset"
                break
                
            # .. or beyond the end:
            if tind_ >= len(self.ds.time):
                out_of_ice = True
                self.end_reason = "Reached end of dataset"
                break

            t_1 = self.ds.time.data[tind_]

            # Find the new position x_(i-1) = x_i - d_t * u_i     
            # Factor 86400/100: (cm/s) -> (m/day)
            x_1 = self.x[-1] - 86400/100*self.u[-1] 
            y_1 = self.y[-1] - 86400/100*self.v[-1] 
            
                          
            # Calculate u, v for the new location of the particle (u_1, v_1)
            # and check whether we are still in ice (is_ice_1)
            is_ice_1, u_1, v_1 = self.get_uv(tind_, x_1, y_1)
                          
            # Append new values to the vectors (t, x, y, u, v)
            self.time_num.append(date2num(t_1))
            self.x.append(x_1)
            self.y.append(y_1)
            self.u.append(u_1)
            self.v.append(v_1)

            #### Print status #############################
                          
            # Net displacement from origin (km)
            displ = np.sqrt((self.x[-1] - self.x[0])**2 
                            + (self.y[-1] - self.y[0])**2)*1e-3
            
            print('BACKTRACKING: Time step %i (%s) - %.1f km ..\r'%(
                nn, pd.Timestamp(t_1).strftime('%d %b %Y'), displ), end = '')
            
            #### Evaluating whether to continue ###########
                          
            # A. Stop if the particle is in open water
            if is_ice_1==False and 'A' in self.end_criteria:
                out_of_ice = True
                print('HIT OPEN WATER! ()')
                self.end_reason = ('(A) Particle hit open water '
                       '(nearest non-landmasked point has u = v = NaN (<15%SIC))')
                          
            # B. Stop if ice hasn't moved for 3 days
            if nn > 3 and 'B' in self.end_criteria:
                u_last = np.asarray(self.u[-3:])
                v_last = np.asarray(self.v[-3:])
                umax_last = np.sqrt(u_last**2 + v_last**2).max()
                if umax_last < 0.00116:
                    out_of_ice = True
                    print('NO MOVEMENT! (Particle movement stalled last 3 days)')
                    self.end_reason = (
                        '(B) No movement over 3 days '
                        '(|u|<100 m/day for 3 consecutive days)'
                    )

            # C. Stop if ice has drifted into land (defined by MASK.is_land)
            if self.is_on_land(x_1, y_1) and 'C' in self.end_criteria:
                out_of_ice = True
                print('LAND AHOY! (Particle hit a coastline)')
                self.end_reason = '(C) Hit "land" (based on land mask)'

            #### (Loop ends when out_of ice = True) ##########    
                          
                          
        ###### Store the final time, location displacement and travel time
        self.date_origin = num2date(self.time_num[-1])
        dx_final = self.x[-1] - self.x[0] 
        dy_final = self.y[-1] - self.y[0] 
        
        self.lon, self.lat = IceParticle.ease2_to_lonlat.transform(self.x, self.y)
        self.x_origin, self.y_origin = self.x[-1], self.y[-1]
        self.lon_origin, self.lat_origin = IceParticle.ease2_to_lonlat.transform(
            self.x[-1], self.y[-1])

        self.total_displacement = np.sqrt(dx_final**2 + dy_final**2)
        self.travel_time_days = date2num(self.time0) - date2num(self.date_origin)

        ### Convert lists to arrays
        for attr in ['time_num', 'x', 'y', 'u', 'v', 'lat', 'lon']:
            setattr(self, attr, np.asarray(getattr(self, attr)))

                          
                          
    # Export to dictionary                      
    def to_dict(self):
        DICT = {}
        for key in ['u', 'v', 'x', 'y', 'lat', 'lon', 'time_num', 
                    'date_origin', 'x_origin', 'y_origin', 
                    'lat0', 'lon0', 'x0', 'y0', 'time0','tind0', 
                    'total_displacement', 'travel_time_days', 'end_reason']:
                          
            if hasattr(self, key):
                DICT[key] = getattr(self, key)
            
        return DICT
```
